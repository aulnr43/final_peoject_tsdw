```python
print("Statistik Deskriptif Numerik Hotels")
display(hotels.describe().T)

print("\nStatistik Deskriptif Kategorikal Hotels")
display(hotels.describe(include="object").T)

ringkasan_kolom = pd.DataFrame({
    "nama_kolom": hotels.columns,
    "pandas_dtype": hotels.dtypes.astype(str).values,
    "jumlah_missing": hotels.isna().sum().values,
    "persen_missing": (hotels.isna().mean() * 100).round(2).values,
    "jumlah_unik": hotels.nunique(dropna=True).values,
    "contoh_nilai": [hotels[col].dropna().unique()[:4].tolist() for col in hotels.columns]
})
display(ringkasan_kolom)
```

```python
gabung = pd.concat([H1, H2], ignore_index=True)
cek = []
for kol_h, kol in zip(gabung.columns, hotels.columns[1:]):
    a, b = gabung[kol_h], hotels[kol]
    if not pd.api.types.is_numeric_dtype(a):
        a = a.str.strip().replace("NULL", np.nan)
    if pd.api.types.is_numeric_dtype(b):
        a = pd.to_numeric(a, errors="coerce")
    sama = (a == b).fillna(False).astype(bool) | (a.isna() & b.isna())
    cek.append((kol, int((~sama).sum())))
cek = pd.DataFrame(cek, columns=["kolom", "baris_berbeda"])
print("Total sel berbeda setelah strip spasi & 'NULL' -> NaN:", cek["baris_berbeda"].sum())
print("Label hotel sesuai urutan H1 lalu H2:",
      (hotels["hotel"].iloc[:len(H1)] == "Resort Hotel").all()
      and (hotels["hotel"].iloc[len(H1):] == "City Hotel").all())
```

```python
reports_dir = Path("reports")
reports_dir.mkdir(parents=True, exist_ok=True)

kamus_data = {
    "hotel": ("Kategorikal", "Jenis hotel: Resort Hotel (H1) atau City Hotel (H2)", "{'Resort Hotel', 'City Hotel'}"),
    "is_canceled": ("Biner / Integer", "Status pembatalan reservasi (1 = dibatalkan, 0 = tidak dibatalkan)", "{0, 1}"),
    "lead_time": ("Integer", "Jumlah hari antara tanggal pemesanan masuk PMS dan tanggal kedatangan", ">= 0"),
    "arrival_date_year": ("Integer", "Tahun tanggal kedatangan tamu", "{2015, 2016, 2017}"),
    "arrival_date_month": ("Kategorikal", "Bulan tanggal kedatangan tamu", "January - December"),
    "arrival_date_week_number": ("Integer", "Nomor minggu dalam tahun pada tanggal kedatangan", "1 - 53"),
    "arrival_date_day_of_month": ("Integer", "Tanggal dalam bulan pada tanggal kedatangan", "1 - 31"),
    "stays_in_weekend_nights": ("Integer", "Jumlah malam akhir pekan (Sabtu/Minggu) tamu menginap atau dipesan", ">= 0"),
    "stays_in_week_nights": ("Integer", "Jumlah malam hari kerja (Senin-Jumat) tamu menginap atau dipesan", ">= 0"),
    "adults": ("Integer", "Jumlah tamu dewasa", ">= 0"),
    "children": ("Float / Integer", "Jumlah tamu anak-anak", ">= 0"),
    "babies": ("Integer", "Jumlah tamu bayi", ">= 0"),
    "meal": ("Kategorikal", "Paket makan yang dipesan (Undefined/SC, BB, HB, FB)", "{'BB', 'FB', 'HB', 'SC', 'Undefined'}"),
    "country": ("Kategorikal", "Kode negara asal tamu (ISO 3166-1 alpha-3)", "Kode ISO 3 huruf (mis. PRT, GBR, FRA)"),
    "market_segment": ("Kategorikal", "Segmen pasar pemesanan (TA = Travel Agents, TO = Tour Operators)", "Online TA, Offline TA/TO, Groups, Direct, Corporate, dll."),
    "distribution_channel": ("Kategorikal", "Saluran distribusi pemesanan", "TA/TO, Direct, Corporate, GDS, Undefined"),
    "is_repeated_guest": ("Biner / Integer", "Penanda tamu berulang (1 = ya, 0 = bukan)", "{0, 1}"),
    "previous_cancellations": ("Integer", "Jumlah pemesanan sebelumnya yang dibatalkan oleh tamu ini", ">= 0"),
    "previous_bookings_not_canceled": ("Integer", "Jumlah pemesanan sebelumnya yang tidak dibatalkan oleh tamu ini", ">= 0"),
    "reserved_room_type": ("Kategorikal", "Kode tipe kamar yang dipesan (dianonimkan)", "Kode huruf A - L"),
    "assigned_room_type": ("Kategorikal", "Kode tipe kamar yang diberikan saat operasional (dianonimkan)", "Kode huruf A - L"),
    "booking_changes": ("Integer", "Jumlah perubahan/amandemen pemesanan sebelum check-in atau batal", ">= 0"),
    "deposit_type": ("Kategorikal", "Jenis jaminan deposit pemesanan", "{'No Deposit', 'Non Refund', 'Refundable'}"),
    "agent": ("Float / ID", "ID agen perjalanan pemesan (NaN jika tanpa agen)", "ID numerik atau NaN"),
    "company": ("Float / ID", "ID perusahaan pemesan/penanggung biaya (NaN jika bukan perusahaan)", "ID numerik atau NaN"),
    "days_in_waiting_list": ("Integer", "Jumlah hari pemesanan di daftar tunggu sebelum dikonfirmasi", ">= 0"),
    "customer_type": ("Kategorikal", "Tipe kontrak/kelompok pelanggan", "{'Contract', 'Group', 'Transient', 'Transient-Party'}"),
    "adr": ("Float", "Average Daily Rate: rata-rata tarif kamar per malam menginap", ">= 0"),
    "required_car_parking_spaces": ("Integer", "Jumlah kebutuhan slot parkir mobil", ">= 0"),
    "total_of_special_requests": ("Integer", "Jumlah permintaan khusus tamu", ">= 0"),
    "reservation_status": ("Kategorikal", "Status akhir reservasi", "{'Canceled', 'Check-Out', 'No-Show'}"),
    "reservation_status_date": ("Tanggal / String", "Tanggal ditetapkannya status terakhir reservasi", "Format YYYY-MM-DD")
}

def ubah(kolom, tipe=None, desk=None, domain=None):
    t, d, e = kamus_data[kolom]
    kamus_data[kolom] = (tipe or t, desk or d, domain or e)

ubah("arrival_date_month", tipe="Kategorikal ordinal")
ubah("agent", tipe="Kategorikal nominal (ID anonim)")
ubah("company", tipe="Kategorikal nominal (ID anonim)")
ubah("children", tipe="Integer (terbaca float karena ada NaN)")
ubah("reservation_status_date", tipe="Tanggal (tersimpan sebagai string)")
ubah("adr", tipe="Numerik kontinu",
     desk="Average Daily Rate: total transaksi penginapan dibagi total malam menginap")
ubah("meal", desk="Paket makan: Undefined/SC = tanpa paket, BB, HB, FB")
ubah("country",
     domain="Seharusnya kode ISO alpha-3; data mentah juga memuat 'CN' (2 huruf)")
ubah("market_segment", domain="{'Aviation', 'Complementary', 'Corporate', 'Direct', "
     "'Groups', 'Offline TA/TO', 'Online TA', 'Undefined'}")
ubah("reserved_room_type", domain="{A, B, C, D, E, F, G, H, L, P}")
ubah("assigned_room_type", domain="{A, B, C, D, E, F, G, H, I, K, L, P}")

rows_dict = []
for col in hotels.columns:
    konsep, desc, exp = kamus_data[col]
    rows_dict.append({
        "variable_name": col,
        "pandas_dtype": str(hotels[col].dtype),
        "variable_type": konsep,
        "missing_count": int(hotels[col].isna().sum()),
        "missing_pct": round(float(hotels[col].isna().mean() * 100), 2),
        "unique_count": int(hotels[col].nunique(dropna=True)),
        "description": desc,
        "expected_values": exp
    })

df_data_dict = pd.DataFrame(rows_dict)
output_path = reports_dir / "data_dictionary.csv"
df_data_dict.to_csv(output_path, index=False)
print("File berhasil disimpan ke:", output_path)
display(df_data_dict)
```

```python
import subprocess
from pathlib import Path
from urllib.parse import quote
from google.colab import userdata

repo_dir = Path("/content/final_peoject_tsdw")
rel_file = "reports/data_dictionary.csv"
github_user = "aulnr43"
repo_name = "final_peoject_tsdw"
branch = "main"

def run_git(*args, check=True):
    p = subprocess.run(
        ["git", "-C", str(repo_dir), *args],
        capture_output=True,
        text=True,
        check=check,
    )
    if p.returncode != 0 and check:
        print("STDOUT:", p.stdout)
        print("STDERR:", p.stderr)
        p.check_returncode()
    return p

# 1. Batalkan rebase/merge yang nyangkut (jika ada) & samakan repo lokal ke origin/main
run_git("rebase", "--abort", check=False)
run_git("merge", "--abort", check=False)
run_git("fetch", "origin")
run_git("reset", "--hard", f"origin/{branch}")

# 2. Pastikan identitas git lokal terisi
if not run_git("config", "user.name", check=False).stdout.strip():
    run_git("config", "user.name", github_user)
if not run_git("config", "user.email", check=False).stdout.strip():
    run_git("config", "user.email", f"{github_user}@users.noreply.github.com")

# 3. Simpan ulang df_data_dict terbaru ke reports/data_dictionary.csv
csv_path = repo_dir / rel_file
csv_path.parent.mkdir(parents=True, exist_ok=True)
df_data_dict.to_csv(csv_path, index=False)

# 4. Commit jika isi CSV berbeda dengan versi di GitHub
if run_git("status", "--porcelain", rel_file).stdout.strip():
    run_git("add", rel_file)
    run_git("commit", "-m", "Update reports/data_dictionary.csv")

    raw_token = userdata.get("GITHUB_TOKEN")
    if not raw_token:
        raise ValueError("Secret 'GITHUB_TOKEN' belum diisi atau akses notebook belum dinyalakan.")

    token = quote(raw_token, safe="")
    auth_url = f"https://{github_user}:{token}@github.com/{github_user}/{repo_name}.git"
    clean_url = f"https://github.com/{github_user}/{repo_name}.git"

    run_git("remote", "set-url", "origin", auth_url)
    try:
        run_git("push", "origin", branch)
        print("Berhasil push reports/data_dictionary.csv terbaru ke GitHub!")
    finally:
        run_git("remote", "set-url", "origin", clean_url, check=False)
else:
    print("Isi reports/data_dictionary.csv sudah sama persis dengan versi terbaru di GitHub.")

print("\nStatus branch sekarang:")
print(run_git("branch", "-vv").stdout.strip())
```
